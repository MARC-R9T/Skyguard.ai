# Notebook 02 — Data Preprocessing
SkyGuard AI | Cleaning, Filtering, Feature Engineering

```python
import sys, os
sys.path.insert(0, os.path.abspath("../backend"))
import pandas as pd
import numpy as np
from src.data_preprocessing import clean_opensky, filter_region
print("Modules loaded")
```

## 1. Load Raw Data

```python
raw_df = pd.read_csv("../backend/data/opensky_log.csv")
print("Raw rows:", len(raw_df))
raw_df.head()
```

## 2. Apply clean_opensky()

```python
cleaned = clean_opensky(raw_df)
print("After cleaning:", len(cleaned), "rows")
print("Dropped:", len(raw_df) - len(cleaned))
cleaned.head()
```

## 3. Apply filter_region()

```python
filtered = filter_region(cleaned)
print("After region filter:", len(filtered), "rows")
```

## 4. Column Normalization Check

```python
print("Columns:", filtered.columns.tolist())
print("Dtypes:")
print(filtered.dtypes)
```

## 5. Duplicate ICAO24 Handling
Each ICAO24 should appear only once (most recent state).

```python
dupes = filtered["icao24"].duplicated().sum()
print("Duplicate icao24 entries:", dupes)
deduped = filtered.groupby("icao24", as_index=False).tail(1)
print("After dedup:", len(deduped))
```

## 6. Null Value Summary After Cleaning

```python
print(deduped.isnull().sum())
```

## 7. Save Cleaned Output

```python
deduped.to_csv("../backend/data/cleaned_data.csv", index=False)
print("Saved cleaned_data.csv")
```
