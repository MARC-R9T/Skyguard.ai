# Notebook 01 — Exploratory Data Analysis
SkyGuard AI | ADS-B + OpenSky Flight Data

```python
import sys, os
sys.path.insert(0, os.path.abspath("../backend"))
import pandas as pd
import numpy as np
import plotly.express as px
print("Libraries loaded")
```

## 1. Load Datasets

```python
adsb_df = pd.read_csv("../backend/data/adsb_cleaned.csv")
opensky_df = pd.read_csv("../backend/data/cleaned_data.csv")
print("ADS-B rows:", len(adsb_df))
print("OpenSky rows:", len(opensky_df))
```

## 2. Dataset Info

```python
adsb_df.info()
```

## 3. Basic Statistics

```python
adsb_df.describe()
```

## 4. Missing Value Analysis

```python
missing = adsb_df.isnull().sum().sort_values(ascending=False)
fig = px.bar(missing[missing > 0], title="Missing Values per Column")
fig.show()
```

## 5. Geographic Distribution

```python
sample_df = adsb_df.dropna(subset=["lat","lon"]).sample(min(5000, len(adsb_df)))
fig = px.scatter_geo(sample_df, lat="lat", lon="lon", color="velocity",
    title="ADS-B Flight Positions (Sample 5k)", projection="natural earth")
fig.show()
```

## 6. Altitude Distribution

```python
fig = px.histogram(adsb_df.dropna(subset=["alt"]), x="alt", nbins=50,
    title="Altitude Distribution (metres)")
fig.show()
```

## 7. Velocity Distribution

```python
fig = px.histogram(adsb_df.dropna(subset=["velocity"]), x="velocity", nbins=50,
    title="Ground Speed Distribution (m/s)")
fig.show()
```

## 8. Heading Distribution

```python
fig = px.histogram(adsb_df.dropna(subset=["heading"]), x="heading", nbins=36,
    title="Heading Distribution (degrees)")
fig.show()
```

## 9. Vertical Rate Analysis

```python
fig = px.histogram(adsb_df.dropna(subset=["vertical_rate"]), x="vertical_rate", nbins=50,
    title="Vertical Rate Distribution (m/s)")
fig.show()
```
